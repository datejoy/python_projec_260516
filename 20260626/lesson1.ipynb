```python
# numpy 可處理陣列運算

a = [1, 2, 3]
print(a + [4, 5])   # [1, 2, 3, 4, 5]

import numpy as np

b = np.array([1, 2, 3])
print(b + np.array([4, 5, 6]))   # [5 7 9]
```

```python
import numpy as np
import math
import matplotlib.pyplot as plt

t = np.arange(0, 2.5, 0.1)
t1 = t * math.pi
y1 = np.sin(t1)
y2 = np.cos(t1)
# print(y1)

# 建立畫布
fig = plt.figure()

# 建立圖表
axes = fig.add_subplot(1, 1, 1)
# 實體方法
axes.plot(t, y1)
axes.plot(t, y2)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

t = np.arange(0, 5, 0.1)
# np 裡面也有pi
y1 = np.sin(t * np.pi)
y2 = np.cos(t * np.pi)

fig = plt.figure()
ax1 = fig.add_subplot(2, 1, 1)
# Axes.plot(*args, scalex=True, scaley=True, data=None, **kwargs)
# *:不限數量的引數值呼叫； ** 不限數量的引數名稱值呼叫 =>可不照順序
ax1.plot(t, y1, linestyle='--', color='red')
ax2 = fig.add_subplot(2, 1, 2)
ax2.plot(t, y2)
ax2.plot(t, y2, color='blue', linestyle='-')

plt.show()
```

```python
# 步驟 1:匯入 numpy,並取別名 np
import numpy as np

# 步驟 2:匯入 matplotlib.pyplot,並取別名 plt
import matplotlib.pyplot as plt

# 步驟 3:用 np.arange() 產生 0 到 4(不含 4)、間隔 0.2 的陣列,存入 x
x = np.arange(0, 4, 0.2)

# 步驟 4:計算 y1 = cos(pi * x)(用 np.cos 與 np.pi)
y1 = np.cos(np.pi * x)

# 步驟 5:計算 y2 = cos(2 * pi * x)
y2 = np.cos(2 * np.pi * x)

# 步驟 6:用 subplot 把版面切成 1 列 2 欄,選第 1 格(左圖)
fig = plt.figure()
axes1 = fig.add_subplot(1, 2, 1)

# 步驟 7:畫出 x 與 y1,使用綠色實線
axes1.plot(x, y1, color='green', linestyle='-')

# 步驟 8:用 subplot 選第 2 格(右圖)
axes2 = fig.add_subplot(1, 2, 2)

# 步驟 9:畫出 x 與 y2,使用洋紅色點線
axes2.plot(x, y2, color='#FF00FF', linestyle='-.')

# 步驟 10:顯示圖形
plt.show()
```
