```python
# 如何清理資料

import pandas as pd

# 讀檔
dataFram = pd.read_csv('各鄉鎮市區人口密度.csv')
# 每一欄型態
# print(dataFram.dtypes)

# 有多少列
print(len(dataFram))
# 資訊
# dataFram.info()         

# 改變欄位名稱
dataFram.columns = ["統計年", "區域別", "年底人口數", "土地面積", "人口密度"]
# 另一改名稱方法： 
# dataFram.rename(columns=[])

#移除第0列
df = dataFram.drop(index=0)

# 清除所有有NaN的列
df1 = df.dropna()

# 移除統建年欄位 axis=1代表欄, axis=0代表列
df2 = df1.drop('統計年', axis=1)

# 將欄的型態改變
# 強制轉換為數字，非數字無法轉的會變成NaN
df2["年底人口數"] = pd.to_numeric(df2["年底人口數"], errors="coerce")

# 依年底人口數據將NaN的列都去掉
df2 = df2.dropna(subset=["年底人口數"])

# 轉為整數、浮點數
df2["年底人口數"] = df2["年底人口數"].astype("int64")
df2["土地面積"] = df2["土地面積"].astype("float")
df2["人口密度"] = df2["人口密度"].astype("int64")
df2.info()
df2
```
