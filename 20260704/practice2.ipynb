```python
class Person:
    def __init__(self, name): 
        # self為建立的實體自己  可以不用有引數傳入
        # self可以不用有東西傳入(規則)，但之後的引數就需要了
        # 實體方法一定要有self

        # 建立一個屬性(attribute)
        self.name = name

    # 自訂實體的method
    def introduce(self):
        print(f"哈囉，我的名字是{self.name}")
```

```python
p2 = Person("JOY")
p2.introduce()
p3 = Person("MAX")
p3.introduce()
```
