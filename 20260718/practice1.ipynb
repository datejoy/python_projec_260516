```python
import numpy as np
from numpy import ndarray
import pandas as pd
from pandas import DataFrame

scores: ndarray | None = None
scores_df: DataFrame | None = None

scores = np.random.randint(50, 101, size=(50, 5))
scores_df = pd.DataFrame(scores,
                         columns = ["國文", "數學", "英文", "地理", "歷史"],
                         index = range(1,51))
# print(scores_df.T)

names_df = pd.read_csv('students.csv')
# 如果裡面沒有姓名就會自動補上
# names_df[]外面是subscript ，裡面是list
scores_df[['姓名','性別']] = names_df[['姓名','性別']].head(50).values
scores_df = scores_df[['姓名','性別', "國文", "數學", "英文", "地理", "歷史"]]
# print(scores_df)

# 選取欄
scores_df[["姓名", "國文"]]
# 選取列
scores_df.iloc[5:10]
```

```python
# 透過值搜尋
# display只有ipynb有 
display(scores_df[scores_df['姓名'] == '許美慧'])

scores_df[scores_df['國文']<60]

scores_df.query("性別 == '女' and 國文 < 60")
```

```python
scores_df["總分"] = scores_df[["國文", "數學", "英文", "地理", "歷史"]].sum(axis=1)
scores_df["平均"] = scores_df[["國文", "數學", "英文", "地理", "歷史"]].mean(axis=1)
# ascending=False 由大到小
scores_df["排名"] = scores_df["總分"].rank(ascending=False, method='min')
scores_df = scores_df.sort_values(by = '排名')
scores_df
```

```python
from pandas import Series

def calc(s:Series)->int:
    n = (s<60).sum()
    return n

scores_df['不及格的科目數量'] = scores_df[['國文', '英文', '數學', '地理', '歷史']].apply(calc,axis=1)
scores_df
```
