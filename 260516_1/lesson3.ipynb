```python
# 學生總分為300
# 有些學生可加分5%

while True:
    try: 
        score = int(input("請輸入學生分數(最高300分): "));
        break; # 跳出迴圈
    except ValueError:
        print("請輸入整數");

isAdd = input("學生是否符合加分條件？(y/n)");
if isAdd == "y": 
    score = score * 1.05;
    #四捨五入
    score = int(score + 0.5);
if(score > 300):
        score = 300;

print("學生分數：", score);
```
