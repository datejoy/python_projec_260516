```python
# for in 迴圈
for i in range(5):
    print(i);

print("===============")

for i in range(1, 5):
    print(i);

print("===============")

for i in range(1, 10, 2):
    print(i);

print("===============")

total = 0
for i in range(1, 101):
    total += i
print(total);
```

```python
score = int(input("請輸入分數："))

if score >= 90:
    print("A")
elif score >= 80:   # else 的條件為上方的false，此行的條件實為 score < 90 && score >= 80
    print("B")
elif score >= 70:   #此行的條件實為 score < 80 && score >=70
    print("C")
elif score >= 60:   #此行的條件實為 score < 70 && score >=60
    print("D")
else:               #此行的條件實為 score < 60
    print("F")
```
